# SISO Adaptive Control — Methodological Notes

Data-driven identification and control of Single-Input Single-Output (SISO) systems using a Linear Neural Unit (LNU), a Quadratic Neural Unit (QNU), see, e.g., [1]–[4], and optionally a Multilayer Perceptron (MLP), with Principal Component Analysis (PCA), Model Reference Adaptive Control (MRAC), and Model Predictive Control (MPC) using machine learning. Physical models (differential equations) are used only as examples of controlled plants; they serve solely to simulate data and the controlled system, while only their measured (simulated) data are used for control design.

## Practical selection of the neural architecture and settings

The emphasis is on the relationships among sampling, history length, the data operating range, recursive operation of the dynamic model, spectral diagnostics, and attainability of the required control performance.

---
Note: The examples do not necessarily represent the best configurations that can be achieved for a given system.


# 1. What Should Be Verified Before Controller Design

A low one-step prediction error alone may not be sufficient. For control use, it is also appropriate to assess:

- the behavior of the dynamic model during recursive multi-step prediction;
- coverage of the operating range of the manipulated variable $u$, which is the only input in the training data;
- coverage of the operating range of the controlled variable $y$ and the trajectory of the desired variable $d$;
- the dynamic stability of the learned neural model, ideally $\rho(\mathbf{A}_y(k))<1$ for LNU and QNU, and $\rho(J_y)<1$ for MLP, for as many data samples as possible, ideally $\forall k$; see, e.g., [1];
- the numerical stability of learning, $\rho(\mathbf{A}_w(k))<1$; see, e.g., [2], [3].


# 2. Operating Ranges of $u$, $d$, and $y$, and Reference Attainability

During identification, the selected ranges of $u$, $d$, and $y$ should be representative of the expected operation.

It is useful to compare the region used for training with the region required during control:

$$
u_{\mathrm{train}},\ d_{\mathrm{train}},\ y_{\mathrm{train}}
\quad \leftrightarrow \quad
u_{\mathrm{control}},\ d_{\mathrm{control}},\ y_{\mathrm{ref}}.
$$

The required reference may be unattainable in amplitude if the available range of $u$ is insufficient, or dynamically unattainable if the reference model requires a response that the plant and actuator cannot realize.

For both MRAC and MPC, attainability of the reference model, i.e., the required behavior of the entire controlled system, is essential. Two forms can be distinguished:

- **amplitude attainability** — whether the controlled system can reach the required value of $y$ at all;
- **dynamic attainability** — whether the controlled system can achieve the required speed, damping, and response shape defined by the reference model or prescribed in another way.

For MRAC, the reference model must be selected so that its dynamics are attainable with respect to the plant dynamics, the limited range of the manipulated variable, and what the controller can learn from the available data and the selected structure. A reference model that is too fast, insufficiently damped, or otherwise unrealistic may lead to a persistent control error, saturation of the manipulated variable, or unstable adaptation.

In MPC, the constraint on $u$ can be included directly in the optimization; however, MPC cannot overcome physical unattainability or an insufficiently learned operating range.


# 3. Selection of the Sampling Period $dt$

The sampling period $dt$ must capture the fastest phenomena relevant to prediction and control.

An excessively large $dt$ may hide fast transients, impair the representation of oscillations, and compress a transport delay into only a few samples.

An excessively small $dt$ may create strongly correlated neighboring samples, require larger $n_u$ and $n_y$, increase the regressor dimension, emphasize noise, and worsen numerical conditioning.

The practical objective is to select the largest $dt$ that still preserves the dynamics relevant to control.


# 4. Selection of $n_u$ and $n_y$

The parameters $n_u$ and $n_y$ determine the lengths of the input and output histories:

$$
T_u = n_u \cdot dt, \qquad T_y = n_y \cdot dt  \ \ \ [\mathrm{sec}].
$$

The history should cover transport delay, relevant dynamic memory, and a substantial part of the oscillatory behavior.

A history that is too short does not contain the required information. A history that is too long increases the number of parameters, collinearity, and the risk of overfitting.

Start with smaller values of $n_u$ and $n_y$, and extend the history only if one-step prediction, recursive multi-step prediction, and control usability improve.


# 5. Linear Neural Unit (LNU)

The LNU should be the default model. LNU and QNU belong to Higher-Order Neural Unit (HONU) architectures whose learning is linear in the adapted parameters [1], [2].

This also applies to many physically nonlinear systems. Within a limited operating range, the input-output dynamics may be approximated sufficiently well by an affine-linear dynamic model.

An LNU trained in batch mode on the entire training dataset is often sufficient for control, even when a QNU provides more accurate identification.

For control, the model should adequately capture the sign and magnitude of the input effect, dynamic memory, delay, the operating ranges of $u$ and $y$, and the behavior during recursive operation.

Higher identification accuracy of a more complex model does not necessarily imply a better control result.


# 6. Quadratic Neural Unit (QNU)

A QNU is appropriate primarily when the LNU does not provide sufficient identification or control quality within the selected operating range. The QNU extends the LNU with quadratic and interaction terms while preserving linearity in the adapted weights [1], [2].

A QNU may be advantageous for asymmetric systems, amplitude-dependent dynamics, nonlinear friction, state-dependent gain, and interactions among delayed inputs and outputs.

For asymmetric models, a QNU may identify the dynamics substantially better than an LNU. For control itself, however, an LNU is often sufficient if it captures the relevant dynamics adequately.

The QNU increases the regressor dimension and may worsen numerical conditioning. Its benefit should therefore be assessed according to recursive prediction and control behavior, not only according to the training error.


# 7. Multilayer Perceptron (MLP)

In MPC, the MLP is used as an alternative predictive plant model.

It is intended primarily for comparison and for cases in which neither the LNU nor the QNU is sufficient. For the tested simulated systems, an MLP was not required to achieve usable control performance.

This is consistent with the fact that control was performed within limited operating ranges, the LNU often captured the local input-output dynamics adequately, and the regressor history represented a substantial part of the system memory.

The MLP has greater approximation capacity, but also higher demands on data, architecture, initialization, validation, and recursive operation.


# 8. Principal Component Analysis (PCA)

In this software, PCA is part of MPC and is used to reduce the predictive regressor. It is not used in MRAC.

Its main purpose is to reduce collinearity and excessive regressor dimensionality. Collinearity arises particularly when:

- $dt$ is small and neighboring samples are very similar;
- $n_u$ or $n_y$ are larger;
- the QNU produces a large number of mutually dependent quadratic and interaction terms.

In MPC, the number of principal components can be selected in two ways:

- **by explained variance** — the smallest number of components required to reach the selected fraction of explained variance is retained;
- **by rank** — the number of components is limited according to the numerical rank of the data regressor.

PCA can:

- remove linearly dependent or nearly dependent directions;
- reduce the effective dimension of the predictive model;
- improve the numerical conditioning of learning and MPC optimization;
- reduce redundancy caused by fine sampling and a longer history;
- substantially reduce the QNU regressor.

In MPC, there is usually no reason to disable PCA automatically, particularly for larger or collinear regressors. It is nevertheless necessary to verify that the selected reduction has not degraded recursive multi-step prediction and has not removed directions important for the effect of the manipulated variable on the output.

PCA does not replace an appropriate selection of $dt$, $n_u$, and $n_y$. If the history does not contain the required dynamic information, PCA cannot supply it.


# 9. One-Step Prediction and Recursive Operation of the Dynamic HONU Model

During one-step prediction, measured past outputs are used in the regressor. The error is therefore corrected by the actual measurement at each step.

During recursive multi-step prediction, predicted outputs are fed back into the regressor. The model is thus operated as a dynamic recurrent HONU model.

A low one-step prediction error does not guarantee accurate multi-step prediction. A small local error may accumulate through the recursion, and the model may be accurate for identification but unsuitable for MPC.


# 10. Meaning of $\rho(A_y)$ and $\rho(J_y)$

The spectral radius $\rho(A_y)$, or $\rho(J_y(k))$, is used to assess the **local dynamic stability of the learning neural architecture**.

For an LNU, $A_y$ is the local dynamic matrix of the neural architecture. For a QNU, the first row of the local dynamic matrix $A_y(k)$ is determined analytically, and its coefficients therefore depend on the current regressor over time. For an MLP, the local Jacobian of the output recursion, $J_y(k)$, obtained by differentiating the network at the operating point, i.e., at time $k$, is used analogously.

$$
\rho(A_y)=\max_i\left|\lambda_i(A_y)\right|.
$$

- $\rho<1$ indicates locally stable recursive output dynamics;
- $\rho\approx1$ indicates marginal or weakly damped behavior;
- $\rho>1$ indicates locally unstable recursive model dynamics.

This implemented diagnostic applies to the learned model and indicates the suitability of the trained neural model for controller training and control verification. By itself, it is not a proof of global stability of the physical plant or the closed control loop [1], [2], but it contributes substantially to practical stability assessment.


# 11. Relationship Between $\rho(A_y)$ and Controller Trainability

$\rho(A_y)$ affects controller trainability indirectly.

If the recursive model is locally expansive, the prediction error, sensitivities with respect to future inputs, gradients of the control criterion, and numerical sensitivity of the optimization may grow during multi-step prediction. This may make training of the MRAC controller and optimization of MPC more difficult.

Conversely, $\rho(A_y)<1$ alone does not guarantee good controller trainability. The difficulty may arise from weak output sensitivity to the input, insufficient excitation, an unsuitable reference model, poor scaling, insufficiently covered operating ranges of $u$, $d$, and $y$, a short history, an unsuitable horizon, or active constraints.


# 12. MRAC — When and Why

The HONU architecture for discrete-time MRAC and its batch and online learning are described in [3]. MRAC is generally computationally simpler and faster than MPC because it does not solve a finite-horizon optimization problem at every sampling step.

MRAC is particularly suitable when an attainable reference model can be defined, explicit constraints on the future trajectory are not decisive, online adaptation is required, and low computational cost is important.

In the online adaptive MRAC branch, the controller continues to learn during control. The controller weights are further updated according to the control error and the current behavior of the system.

Attainability of the reference model must be assessed with respect to the range of $u$, the disturbance range $d$, the operating range of $y$, and the dynamic capabilities of the plant and actuator.

# 13. MPC — When and Why

MPC is more appropriate when it is necessary to work explicitly with transport delay, the prediction horizon, constraints on $u$, constraints on the rate of change of $u$, a future reference, or a trade-off between reference tracking and control effort.

MPC supports LNU, QNU, and MLP predictive models.

MPC is more flexible than MRAC, but at the cost of higher computational demand and a larger number of tuning parameters. A long horizon is not automatically advantageous; for a recursively sensitive model, it may worsen both the optimization and the control result.

# 14. MRAC versus MPC — Practical Selection

**Prefer MRAC** when simplicity, speed, online adaptation, and the existence of a realistically attainable reference model are priorities.

**Prefer MPC** when delay, constraints on $u$, a future reference, or an explicit trade-off between control error and control effort are essential.

The physical class of the system does not necessarily determine the choice. The decisive factors are reference attainability, coverage of the operating ranges of $u$, $d$, and $y$, the significance of delay $\tau_u$, actuator constraints, the quality of the recursive learned model, and the available computational power.

The selected results illustrate the use of MRAC for servo-mechanical, biomedical, ecological, and process systems, and the use of MPC particularly for delayed mechanical, network, and power systems.


# 15. Bias Terms $x_0$ and $\xi_0$

The bias term $x_0$ in the plant model and the bias term $\xi_0$ in the controller allow affine rather than purely linear mappings.

For an LNU plant model:

$$
\hat y(k) = w_0 + \mathbf{w}^{\mathsf T}\mathbf{x}(k).
$$

For an LNU controller:

$$
u(k) = v_0 + \mathbf{v}^{\mathsf T}\boldsymbol{\xi}(k).
$$

The bias allows representation of a nonzero operating point, a static offset, an asymmetric operating range, a nonzero steady-state component of the manipulated variable, and compensation of a constant disturbance.

The bias itself is not an integrator. In MRAC, however, the controller bias $\xi_0$ may allow the controller to learn a constant steady-state correction that helps reduce the steady-state control error.


# 16. Role of the P Controller

A P controller can form a simple basic stabilization layer.

It can react rapidly to the control error, increase effective damping, limit deviations from the operating point, restrict excursions into strongly nonlinear regions, facilitate the operation of MRAC or MPC, and serve as a diagnostic reference baseline.

A P controller does not remove plant nonlinearities, but it can reduce their practical influence by keeping the system within a narrower operating range.

In combination with MRAC or MPC, the P branch can stabilize the basic dynamics, while the adaptive or predictive component generates an additional control action. This combination can reduce the steady-state control error, particularly if the adaptive or predictive component can generate a nonzero steady-state component of $u$.

A P controller alone does not generally guarantee elimination of the steady-state error.


# 17. Extension to PI or PID

The basic stabilization branch can be extended from a P controller to a PI or PID controller.

- The **P component** provides an immediate stabilizing response.
- The **I component** accumulates the control error and may eliminate the steady-state control error.
- The **D component** may increase damping of fast and oscillatory transients.

MRAC or MPC can then operate on top of the basic PI/PID stabilization layer and improve the dynamic behavior within the learned operating range.

When PI/PID control is used, anti-windup under saturation, filtering of the derivative component, the effect of delay, and interaction between the adaptive/predictive and basic control branches must be considered.

Version 1.0.0 implements only a P controller.


# 18. Recommended Workflow

1. Define the expected ranges of $u$, $d$, and $y$.
2. Select $dt$ according to the fastest relevant phenomenon.
3. Set short embedding histories $n_u$ and $n_y$ (approximately 3–10).
4. Train an LNU in batch mode in MRAC or MPC; in MPC, verify the suitability of the PCA setting.
5. Evaluate the training accuracy of the neural plant model.
7. Check $\rho(A_y)$ or $\rho(J_y)$ over the entire time trajectory (ideally $<1$).
8. Extend the embedding histories only if there is a demonstrable improvement.
9. In MPC, configure PCA according to explained variance or regressor rank and verify prediction after reduction.
10. Try a QNU or MLP if the LNU is systematically insufficient.
11. Use the MLP for comparison or when a demonstrated need exists.
12. Verify attainability of the reference model.
13. In MPC, configure the constraints on $u$, the horizon, and the penalties.
14. Compare the result with the applied P controller.


# 19. Summary

For most tasks, it is appropriate to start with a simple configuration: a reasonable $dt$, a short history, and an LNU trained in batch mode; in MPC, subsequently verify a variant with PCA and an operating range of the training data corresponding to the expected control operation.

A QNU should be used mainly when the LNU is demonstrably insufficient. The MLP is suitable primarily for comparison or for cases requiring greater nonlinear approximation capacity.

MRAC is faster and simpler, but requires an attainable reference model. MPC is more appropriate for delays, constraints, and explicit use of a future horizon to obtain smoother trajectories of the controlled variable.

A P/PI/PID branch can form a basic stabilization layer, while MRAC or MPC improves the system behavior.

No single metric, including $\rho(A_y)$, replaces the joint assessment of identification, recursive dynamics, the operating range, reference attainability, and the control result.
