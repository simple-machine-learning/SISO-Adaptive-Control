# SISO Adaptive Control — metodologické poznámky

Datově řízená identifikace a řízení systémů s jedním vstupem a jedním výstupem (SISO) pomocí lineární neuronové jednotky (LNU), kvadratické neuronové jednotky (QNU), viz např. [1-4], a případně vícevrstvé perceptron sítě (MLP), s využitím metody hlavních komponent (PCA), řízení s referenčním modelem (MRAC) a prediktivním řízením (MPC) se strojovým učením. Fyzikální modely (diferenciální rovnice) jsou využity jen jako příklady řízených soustav; slouží jen k simulaci dat a simulaci řízeného systému a k návrhu řízení jsou použita jen jejich měřená (simulovaná) data.

## Praktická volba neuronového architektury a nastavení

Důraz je kladen na vztah mezi vzorkováním, délkou historie, pracovní oblastí dat, rekurzivním provozem dynamického modelu, spektrálními diagnostikami a dosažitelností požadované regulace.

---
Pozn.: Ukázky nutně nereprezentují nejlepší možné nalezené konfgurace které je možno pro daný systém dosáhnout.

# 1. Co musí být ověřeno před návrhem regulátoru

Nízká chyba jednokroková predikce sama o sobě nemusí stačit. Pro použití je vhodné posoudit i
- chování dynamického modelu při rekurzivní vícekrokové predikci;
- pokrytí pracovní oblasti akční veličiny $u$ která je jediným vstupem v tréninkových datech;
- pokrytí pracovního rozsahu regulované veličiny $y$ a průběhu žádané veličiny $d$;
- dynamickou stabilitu naučeného neuronovéh modelu, ideálně $\rho (\textbf{A}_y(k))<1$ pro LNU a QNU, a $\rho(J_y)<1$ pro MLP pro co nejvíce datových vzorků, ideálně $ \forall k$  viz např. [1];
- numerickou stabilitu učení $\rho (\textbf{A}_w(k))<1$,, viz např. [2,3].

# 2. Pracovní oblast $u$, $d$, $y$ a dosažitelnost reference

Při identifikaci je vhodné mít zvolený rozsah $u$, $d$ a $y$ reprezentativní pro očekávaný provoz.

Je vhodné porovnat oblast použitou při trénování s oblastí požadovanou při regulaci:

$$
u_{\mathrm{train}},\ d_{\mathrm{train}},\ y_{\mathrm{train}}
\quad \leftrightarrow \quad
u_{\mathrm{control}},\ d_{\mathrm{control}},\ y_{\mathrm{ref}}.
$$

Požadovaná reference může být nedosažitelná amplitudově, pokud dostupný rozsah $u$ nestačí, nebo dynamicky, pokud referenční model požaduje odezvu, kterou soustava a aktuátor nemohou realizovat.

U MRAC i MPC je dosažitelnost referenčního modelu, tj. požadovaného chování celého řízeného systému zásadní. Lze rozlišovat:
- **amplitudovou dosažitelnost** — zda regulovaný systém může vůbec dosáhnout požadované hodnoty $y$;
- **dynamickou dosažitelnost** —  zda regulovaný systém může dosáhnout požadované rychlosti, tlumení a tvaru odezvy definované referenčním modelem nebo jinak předepisovaným způsobem.

U MRAC musí být referenční model zvolen tak, aby jeho dynamika byla dosažitelná vzhledem k dynamice soustavy, omezenému rozsahu akční veličiny a tomu, co se regulátor může z dostupných dat a zvolené struktury naučit. Příliš rychlý, málo tlumený nebo jinak nerealistický referenční model může vést k trvalé regulační chybě, saturaci akční veličiny nebo nestabilní adaptaci.

U MPC lze omezení $u$ zahrnout přímo do optimalizace, ale ani MPC nemůže překonat fyzikální nedosažitelnost nebo nedostatečně naučenou pracovní oblast.

# 3. Volba vzorkovací periody $dt$

Vzorkovací perioda $dt$ musí zachytit nejrychlejší jevy relevantní pro predikci a řízení.

Příliš velké $dt$ může skrýt rychlé přechody, zhoršit reprezentaci kmitání a sloučit dopravní zpoždění do několika málo vzorků.

Příliš malé $dt$ může vytvářet silně korelované sousední vzorky, vyžadovat větší $n_u$ a $n_y$, zvětšit dimenzi regresoru, zvýraznit šum a zhoršit numerickou podmíněnost.

Praktickým cílem je zvolit největší $dt$, které ještě zachovává dynamiku významnou pro regulaci.


# 4. Volba $n_u$ a $n_y$

Parametry $n_u$ a $n_y$ určují délku historie vstupu a výstupu:

$$
T_u = n_u \cdot dt, \qquad T_y = n_y \cdot dt  \ \ \ [sec].
$$

Historie má pokrýt dopravní zpoždění, relevantní dynamickou paměť a podstatnou část kmitavého průběhu.

Příliš krátká historie neobsahuje potřebnou informaci. Příliš dlouhá historie zvyšuje počet parametrů, kolinearitu a riziko přeučení.

Začněte menšími hodnotami $n_u$ a $n_y$ a historii prodlužujte pouze tehdy, pokud se zlepšuje jednokroková predikce, rekurzivní vícekroková predikce a regulační použitelnost.


# 5. Linear Neural Unit (LNU)

LNU má být výchozí model. LNU a QNU patří mezi neuronové architektury (Higher-Order Neural Units, HONU), jejichž učení je lineární v adaptovaných parametrech [1], [2].

To platí i pro mnoho fyzikálně nelineárních systémů. V omezené pracovní oblasti může být vstupně-výstupní dynamika dostatečně dobře aproximovatelná afinně-lineárním dynamickým modelem.

LNU naučená dávkovou metodou na celém tréninkovém souboru často pro řízení postačuje, i když QNU poskytuje přesnější identifikaci.

Pro řízení je důležité, aby model dostatečně zachytil znaménko a velikost účinku vstupu, dynamickou paměť, zpoždění, pracovní rozsah $u$ a $y$ a chování při rekurzivním provozu.

Vyšší přesnost identifikace složitějšího modelu nemusí znamenat lepší regulační výsledek.


# 6. Quadratic Neural Unit (QNU)

QNU je vhodné použít především tehdy, když LNU neposkytuje dostatečnou kvalitu identifikace nebo regulace ve zvolené pracovní oblasti. QNU rozšiřuje LNU o kvadratické a interakční členy při zachování linearity v adaptovaných vahách [1], [2].

QNU může být výhodná u asymetrických systémů, amplitudově závislé dynamiky, nelineárního tření, stavově závislého zesílení a interakcí mezi zpožděnými vstupy a výstupy.

U nesymetrických modelů může QNU identifikovat výrazně lépe než LNU. Pro samotné řízení však často postačuje LNU, pokud dostatečně zachycuje relevantní dynamiku.

QNU zvyšuje dimenzi regresoru a může zhoršit numerickou podmíněnost. Její přínos je nutné posuzovat podle rekurzivní predikce a regulačního chování, nikoli pouze podle tréninkové chyby.


# 7. Multilayer Perceptron (MLP)

MLP je v MPC používán jako alternativní prediktivní model rostliny.

Je vhodný zejména pro srovnání a pro případy, kdy LNU ani QNU nestačí. U testovaných simulovaných systémů nebyl MLP pro dosažení použitelné regulace nutný.

To odpovídá tomu, že regulace probíhala v omezených pracovních oblastech, LNU často dostatečně zachytila lokální vstupně-výstupní dynamiku a historie v regresoru zachytila významnou část paměti systému.

MLP má vyšší aproximační kapacitu, ale také vyšší nároky na data, architekturu, inicializaci, validaci a rekurzivní provoz.


# 8. Principal Component Analysis (PCA)

PCA je v tomto softwaru součástí MPC a používá se pro redukci prediktivního regresoru. V MRAC použita není.

Její hlavní význam spočívá v omezení kolinearity a nadměrné dimenze regresoru. Kolinearita vzniká zejména tehdy, když:

- je $dt$ malé a sousední vzorky jsou velmi podobné;
- jsou $n_u$ nebo $n_y$ větší;
- QNU vytváří velké množství vzájemně závislých kvadratických a interakčních členů.

V MPC lze počet hlavních komponent volit dvěma způsoby:

- **podle vysvětlené variability** — zachová se nejmenší počet komponent potřebný k dosažení zvoleného podílu variability;
- **podle hodnosti (hodnosti)** — počet komponent se omezí podle numerické hodnosti datového regresoru.

PCA může:

- odstranit lineárně závislé nebo téměř závislé směry;
- snížit efektivní dimenzi prediktivního modelu;
- zlepšit numerickou podmíněnost učení a optimalizace MPC;
- omezit redundanci způsobenou jemným $dt$ a delší historií;
- výrazně zmenšit QNU regresor.

V MPC obvykle není důvod PCA automaticky vypínat, zejména u rozsáhlejších nebo kolineárních regresorů. Je však nutné ověřit, že zvolená redukce nezhoršila rekurzivní vícekrokovou predikci a neodstranila směry důležité pro účinek akční veličiny na výstup.

PCA nenahrazuje správnou volbu $dt$, $n_u$ a $n_y$. Pokud historie neobsahuje potřebnou dynamickou informaci, PCA ji nemůže doplnit.


# 9. Jednokroková predikce a rekurzivní provoz dynamického HONU modelu

Při jednokrokovou predikci jsou v regresoru použity měřené minulé výstupy. Chyba se proto v každém kroku koriguje skutečným měřením.

Při rekurzivní vícekrokové predikci jsou predikované výstupy vraceny zpět do regresoru. Model je tak provozován jako dynamický rekurentní HONU model.

Nízká chyba jednokrokové predikce nezaručuje kvalitní vícekrokovou predikci. Malá lokální chyba se může v rekurzi akumulovat a model může být přesný při identifikaci, ale nevhodný pro MPC.


# 10. Význam $\rho(A_y)$ a $\rho(J_y)$

Spektrální poloměr $\rho(A_y)$, případně $\rho(J_y(k))$, slouží k posouzení **lokální dynamické stability učící se neuronové architektury**.

Pro LNU je $A_y$ lokální matice dynamiky neuronové architektury. U QNU se první řádek lokální matice dynamiky $A_y(k)$ určuje analyticky a její koeficienty v čase proto závisejí na aktuálním regresoru. U MLP se analogicky používá lokální Jacobian výstupní rekurze $J_y(k)$ získaný derivací sítě v pracovním bodě, tj. v čase $k$.

$$
\rho(A_y)=\max_i\left|\lambda_i(A_y)\right|.
$$

- $\rho<1$ indikuje lokálně stabilní rekurzivní výstupní dynamiku;
- $\rho\approx1$ indikuje hraniční nebo slabě tlumené chování;
- $\rho>1$ indikuje lokálně nestabilní rekurzivní dynamiku modelu.

Tato implementovaná diagnostika se vztahuje k naučenému modelu a naznačuje vhodnost natrénovaného neuronového modelu k nátrénování a ověření řízení. Sama o sobě není důkazem globální stability fyzikální soustavy ani uzavřené regulační smyčky [1], [2], ale významně k posouzení stability prakticky přispívá.


# 11. Vztah $\rho(A_y)$ k natrénovatelnosti regulace

$\rho(A_y)$ ovlivňuje natrénovatelnost regulace nepřímo.

Pokud je rekurzivní model lokálně expanzivní, mohou při vícekrokové predikci růst predikční chyba, sensitivities vůči budoucím vstupům, gradienty regulačního kritéria a numerická citlivost optimalizace. To může ztížit učení MRAC regulátoru i optimalizaci MPC.

Naopak $\rho(A_y) < 1$ samo o sobě nezaručuje dobrou natrénovatelnost regulátoru. Problém může být ve slabé citlivosti výstupu na vstup, nedostatečné excitaci, nevhodném referenčním modelu, špatném škálování, nepokryté pracovní oblasti $u$, $d$ a $y$, krátké historii, nevhodném horizontu nebo aktivních omezeních.


# 12. MRAC — kdy a proč

HONU architektura pro diskrétní MRAC a její dávkové i průběžné učení jsou popsány v [3]. MRAC je obecně výpočetně jednodušší a rychlejší než MPC, protože v každém vzorkovacím kroku neřeší optimalizační úlohu na konečném horizontu.

MRAC je vhodný zejména tehdy, když lze definovat dosažitelný referenční model, nejsou rozhodující explicitní omezení budoucího průběhu, je požadována online adaptace a je důležitá nízká výpočetní náročnost.

U online adaptivní MRAC větve se regulátor během řízení průběžně učí. Váhy regulátoru se dále aktualizují podle regulační chyby a aktuálního chování systému.

Dosažitelnost referenčního modelu musí být posouzena vzhledem k rozsahu $u$, rozsahu poruchy $d$, pracovní oblasti $y$ a dynamickým možnostem soustavy a aktuátoru.

# 13. MPC — kdy a proč

MPC je vhodnější tehdy, když je nutné explicitně pracovat s dopravním zpožděním, predikčním horizontem, omezením $u$, omezením rychlosti změny $u$, budoucí referencí nebo kompromisem mezi sledováním reference a akčním zásahem.

MPC podporuje prediktivní modely LNU, QNU a MLP.

MPC je flexibilnější než MRAC, ale za cenu vyšší výpočetní náročnosti a většího počtu parametrů k ladění. Dlouhý horizont není automaticky výhodný; u rekurzivně citlivého modelu může zhoršit optimalizaci i regulační výsledek.

# 14. MRAC versus MPC — praktické rozhodnutí

**MRAC** preferujte, když je prioritou jednoduchost, rychlost, online adaptace a existuje realisticky dosažitelný referenční model.

**MPC** preferujte, když jsou zásadní zpoždění, omezení $u$, budoucí reference nebo explicitní kompromis mezi regulační chybou a akčním zásahem.

Fyzikální třída systému sama o sobě volbu nemusí určovat. Rozhodující je dosažitelnost reference, pokrytí pracovní oblasti $u$, $d$ a $y$, význam zpoždění $\tau_u$, omezení aktuátoru, kvalita rekurzivního naučeného modelu a dostupný výpočetní výkon.

Vybrané výsledky ilustrují použití MRAC u servomechanismů, biomedicínských, ekologických a procesních systémů a použití MPC zejména u zpožděných mechanických, síťových a energetických systémů.


# 15. Bias členy $x_0$ a $\xi_0$

Bias člen $x_0$ v modelu soustavy a bias člen $\xi_0$ v regulátoru umožňují afinní, nikoli pouze čistě lineární mapování.

U LNU modelu soustavy:

$$
\hat y(k) = w_0 + \mathbf{w}^{\mathsf T}\mathbf{x}(k).
$$

U LNU regulátoru:

$$
u(k) = v_0 + \mathbf{v}^{\mathsf T}\boldsymbol{\xi}(k).
$$

Bias umožňuje reprezentovat nenulový pracovní bod, statický offset, asymetrickou pracovní oblast, nenulovou ustálenou složku akčního zásahu a kompenzaci konstantní poruchy.

Bias sám o sobě není integrátor. U MRAC však bias regulátoru $\xi_0$ může umožnit naučení konstantní ustálené korekce, která pomáhá zmenšit trvalou regulační odchylku.


# 16. Úloha P-regulátoru

P-regulátor může tvořit jednoduchou základní stabilizační vrstvu.

Může rychle reagovat na regulační chybu, zvýšit efektivní tlumení, omezit odchylky od pracovního bodu, omezit pohyb do silně nelineárních oblastí, usnadnit činnost MRAC nebo MPC a sloužit jako diagnostický referenční základ.

P-regulátor nelinearity neodstraňuje, ale může omezit jejich praktický vliv tím, že systém udržuje v užší pracovní oblasti.

Ve spojení s MRAC nebo MPC může P-větev stabilizovat základní dynamiku, zatímco adaptivní nebo prediktivní část vytváří doplňkový akční zásah. Tato kombinace může zmenšit trvalou regulační odchylku, zejména pokud adaptivní nebo prediktivní část dokáže vytvořit nenulovou ustálenou složku $u$.

Samotný P-regulátor odstranění trvalé odchylky obecně nezaručuje.


# 17. Rozšíření na PI nebo PID

Základní stabilizační větev lze rozšířit z P-regulátoru na PI nebo PID.

- **P složka** poskytuje okamžitou stabilizační reakci.
- **I složka** akumuluje regulační chybu a může odstranit trvalou regulační odchylku.
- **D složka** může zvýšit tlumení rychlých a kmitavých přechodů.

MRAC nebo MPC pak může pracovat nad základní PI/PID stabilizační vrstvou a zlepšovat dynamické chování v rámci naučené pracovní oblasti.

Při použití PI/PID je nutné zohlednit anti-windup při saturaci, filtraci derivační složky, vliv zpoždění a interakci mezi adaptivní/prediktivní a základní regulační větví.

Verze v1.0.0 implementuje pouze P-regulátor.


# 18. Doporučený pracovní postup

1. Vymezit očekávaný rozsah $u$, $d$ a $y$.
2. Zvolit $dt$ podle nejrychlejšího relevantního jevu.
3. Nastavit malé historie embedingů $n_u$ a $n_y$ (cca 3-10).
4. Natrénovat dávkově naučenou LNU; v MRAC nebo MPC,  v MPC ověřit vhodnost nastavení PCA.
5. Vyhodnotit přesnost naučení neuronového modelu soustavy.
7. Zkontrolovat $\rho(A_y)$ nebo $\rho(J_y)$ v celém časovém průběhu (ideálně <1).
8. Prodloužit historii embedingů pouze při prokazatelném zlepšení.
9. V MPC nastavit PCA podle vysvětlené variability nebo podle hodnosti regresoru a ověřit predikci po redukci.
10. Zkusit QNU nebo MLP, pokud LNU systematicky nestačí.
11. MLP použít pro srovnání nebo při prokázané potřebě.
12. Ověřit dosažitelnost referenčního modelu.
13. U MPC nastavit omezení $u$, horizont a penalizace.
14. Porovnat výsledek s aplikovaným P-regulátorem.


# 19. Shrnutí

Pro většinu úloh je vhodné začít jednoduchou konfigurací: rozumné $dt$, krátká historie, dávkově naučenou LNU; v MPC následně ověřit variantu s PCA a pracovní oblast tréninkových dat odpovídající očekávanému řízení.

QNU má být použita hlavně tehdy, když LNU prokazatelně nestačí. MLP je vhodný zejména pro srovnání nebo pro případy vyžadující vyšší nelineární aproximační kapacitu.

MRAC je rychlejší a jednodušší, ale vyžaduje dosažitelný referenční model. MPC je vhodnější při zpoždění, omezeních a explicitní práci s budoucím horizontem pro ustálenější průběhy regulované veličiny.

P/PI/PID větev může tvořit základní stabilizační vrstvu, zatímco MRAC nebo MPC zlepšuje chování systému.

Žádná jediná metrika, včetně $\rho(A_y)$, nenahrazuje společné posouzení identifikace, rekurzivní dynamiky, pracovní oblasti, dosažitelnosti reference a regulačního výsledku.
